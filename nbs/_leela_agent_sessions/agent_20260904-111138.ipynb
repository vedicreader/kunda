```
from pathlib import Path
import json
nbs_root = Path('/Users/71293/code/personal/orgs/dhrishti/nbs')
nb_paths = sorted(nbs_root.glob('*.ipynb'))
nb_inventory = {}
for nb_path in nb_paths:
    nb_data = json.loads(nb_path.read_text())
    nb_inventory[nb_path.name] = [
        {'index': i, 'type': cell.get('cell_type'), 'source': ''.join(cell.get('source', [])), 'execution_count': cell.get('execution_count')}
        for i, cell in enumerate(nb_data.get('cells', []))
        if cell.get('cell_type') in ('markdown', 'raw', 'code')
    ]
[(name, len(cells)) for name, cells in nb_inventory.items()]
```

```
for nb_name, cells in nb_inventory.items():
    print(f'\n===== {nb_name} =====')
    for cell in cells:
        source = cell['source'].replace('\n', '\\n')
        print(f"[{cell['index']:02d}] {cell['type']}: {source[:1200]}")
```

```
for nb_name, cells in nb_inventory.items():
    print(f'\n===== {nb_name} PROSE =====')
    for cell in cells:
        if cell['type'] in ('markdown','raw'):
            print(f"\n--- cell {cell['index']} ({cell['type']}) ---\n{cell['source']}")
```

```
for nb_name in ['01_runtime.ipynb','02_serving.ipynb']:
    print(f'\n===== {nb_name} PROSE =====')
    for cell in nb_inventory[nb_name]:
        if cell['type'] in ('markdown','raw'):
            print(f"\n--- {cell['index']} ---\n{cell['source']}")
```
